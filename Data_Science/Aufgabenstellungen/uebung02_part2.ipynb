## 2.2 - Pandas

Laden Sie den *Titanic* Datensatz von https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv

Sie können den Datensatz entweder herunterladen und mit Pandas einlesen oder direkt mit Pandas laden.

```python
import pandas as pd

data = pd.read_csv("../data/titanic.csv")

data
```

Geben Sie die Namen der Spalten aus.

```python
cols = data.columns.to_list()
```

Geben Sie die ersten 5 Zeilen des DataFrames aus.

```python
data.head()
```

Zählen Sie die fehlenden Werte in jeder Spalte des DataFrames und geben die Anzahl aus

```python
for col in cols:
    count = 0 
    for item in data[col]:
        if pd.isna(item):
            count += 1
    print(f'Missing values in {col} is {count}')
```

Entfernen Sie alle Zeilen, die fehlende Werte enthalten, und vergleichen Sie die Form des Datensatzes vor und nach der Bereinigung.

```python
data_clean = data.dropna()

data_clean
```

Füllen Sie die fehlenden Werte in der Spalte "Age" mit dem mittleren Alter auf und überprüfen das Ergebnis.

```python
import numpy as np
mean_age = np.mean(data["Age"])

data["Age"] = [mean_age if pd.isna(age) else age for age in data["Age"]]
```

Konvertieren Sie die Einträge der Spalte "Survived" von `int` in `bool`. 

```python
data["Survived"] = data["Survived"].astype(bool)
data["Survived"].dtype 
```

Benennen Sie die Spalten um, indem Sie alle Groß- durch Kleinbuchstaben austauschen.

```python
cols_lower = [col.lower() for col in cols]

data.columns = cols_lower

data
```
